**Question 1:** What is Data Loading in Power BI and why is it considered the first step of analysis?

**Answer:**

Data Loading in Power BI is the process of importing data from various sources such as Excel files, CSV files, databases, and web services into Power BI. It is considered the first step of analysis because data must be available in Power BI before it can be transformed, modeled, visualized, or analyzed. Proper data loading ensures that accurate and complete information is available for reporting. Without loading data, no analysis or dashboard creation is possible. Therefore, data loading serves as the foundation of the entire Power BI workflow.

**Question 2:** Explain the difference between “Load” and “Transform Data” in Power BI.

**Answer:**

The Load option imports data directly into Power BI without making any changes. It is used when the data is already clean and ready for analysis. The Transform Data option opens Power Query Editor, where users can clean, filter, modify, and shape data before loading it. Transform Data helps remove duplicates, handle missing values, and change data types. Load focuses on importing data, while Transform Data focuses on preparing data for accurate analysis.

**Question 3:** What is a Fact Table and a Dimension Table? Give examples from the dataset.

**Answer:**

A Fact Table contains measurable business data such as confirmed cases, recovered cases, deaths, and active cases. It stores numerical values that are used for calculations and analysis. A Dimension Table contains descriptive information such as Country, State, Date, and Vaccination Status. Fact tables provide quantitative data, while dimension tables provide context for analysis. In the COVID-19 dataset, Confirmed Cases and Deaths can be part of a Fact Table, while Country and Date can be part of Dimension Tables.

**Question 4:** Why is Star Schema preferred over Snowflake Schema in Power BI?

**Answer:**

Star Schema is preferred in Power BI because it has a simple structure with one central fact table connected directly to multiple dimension tables. This design improves query performance and makes data modeling easier to understand. It reduces the number of joins required, which speeds up report generation. Star Schema is also easier to maintain and troubleshoot compared to Snowflake Schema. Therefore, it is widely recommended for Power BI dashboards and reporting solutions.
